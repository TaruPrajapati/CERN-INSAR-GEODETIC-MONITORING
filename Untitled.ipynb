```python
%matplotlib notebook
import h5py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load data from timeseries.h5
with h5py.File('timeseries.h5', 'r') as f:
    ts_data = f['timeseries'][:] * 100  # convert m to cm
    dates = [d.decode('utf-8') for d in f['date'][:]]

dates_dt = pd.to_datetime(dates, format='%Y%m%d')
cum_disp = ts_data[-1, :, :]

# 2. Set up interactive plots
fig, (ax_map, ax_ts) = plt.subplots(1, 2, figsize=(14, 6))

im = ax_map.imshow(cum_disp, cmap='jet', vmin=-5, vmax=5)
ax_map.set_title('Click ANY point on the map!')
ax_map.set_xlabel('Column (X)')
ax_map.set_ylabel('Row (Y)')
plt.colorbar(im, ax=ax_map, label='Cumulative Disp (cm)', shrink=0.8)

point_marker, = ax_map.plot([], [], 'k*', markersize=12, markeredgecolor='white')
ts_line, = ax_ts.plot([], [], 'o-', color='#d62728', linewidth=2, markersize=5)

ax_ts.axhline(0, color='black', linestyle='--', alpha=0.5)
ax_ts.set_title('Displacement Time Series')
ax_ts.set_xlabel('Date')
ax_ts.set_ylabel('Displacement (cm)')
ax_ts.grid(True, linestyle=':', alpha=0.6)
fig.autofmt_xdate()

# 3. Interactive click function
def onclick(event):
    if event.inaxes == ax_map:
        col = int(round(event.xdata))
        row = int(round(event.ydata))
        
        rows, cols = cum_disp.shape
        if 0 <= row < rows and 0 <= col < cols:
            point_marker.set_data([col], [row])
            ax_map.set_title(f'Selected Pixel: Col {col}, Row {row}')
            
            pixel_ts = ts_data[:, row, col]
            ts_line.set_data(dates_dt, pixel_ts)
            
            ax_ts.relim()
            ax_ts.autoscale_view()
            ax_ts.set_title(f'Time-Series for Pixel (Col: {col}, Row: {row})')
            fig.canvas.draw_idle()

fig.canvas.mpl_connect('button_press_event', onclick)
plt.tight_layout()
plt.show()
```
